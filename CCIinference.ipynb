```python
#Apply a CCI inference method on spatial Visium datasets with COMMOT

#import libraries
import commot as ct
import scanpy as sc
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


```

```python
# appeler data spatial deconvolutionnée
adata= 

df_cellchat = ct.pp.ligand_receptor_database(species='human' database='CellChat')

print(df_cellchat.shape)
```

```python
#construction des réseaux de communication
ct.tl.spatial_communication(adata,
    database_name='cellchat', df_ligrec=df_cellchat, dis_thr=500 , heteromeric=True)

```

```python
# plot des graphes de réseaux d'inférences

# spatial direction of a signalling pathway: ex PSAP (à modifier)
ct.tl.communication_direction(adata, database_name='cellchat', pathway_name='PSAP', k=5)
ct.pl.plot_cell_communication(adata, database_name='cellchat', pathway_name='PSAP', plot_method='grid', background_legend=True,
    scale=0.00003, ndsize=8, grid_density=0.4, summary='sender', background='image', clustering='leiden', cmap='Alphabet',
    normalize_v = True, normalize_v_quantile=0.995)

```

```python
# visualiser le niveau de signal envoyé et reçu d'une paire ligand-recepteur (ex: Fgf1-Fgfr1)

pts = adata.obsm['spatial']
s = adata.obsm['commot-cellchat-sum-sender']['s-Fgf1-Fgfr1']
r = adata.obsm['commot-cellchat-sum-receiver']['r-Fgf1-Fgfr1']

fig, ax = plt.subplots(1,2, figsize=(10,4))
ax[0].scatter(pts[:,0], pts[:,1], c=s, s=5, cmap='Blues')
ax[0].set_title('Sender')
ax[1].scatter(pts[:,0], pts[:,1], c=r, s=5, cmap='Reds')
ax[1].set_title('Receiver')
plt.show()
```

```python
#visualiser les directions du signal comme un champ vectoriel

ct.tl.communication_direction(adata, database_name='cellchat', lr_pair=('Fgf1','Fgfr1'), k=5)

ct.pl.plot_cell_communication(adata, database_name='cellchat', lr_pair=('Fgf1','Fgfr1'),
                             plot_method='grid', background_legend=True,
                             scale=0.00003, ndsize=8, grid_density=0.4,
                             summary='sender', background='image', cmap='Alphabet',
                             normalize_v=True, normalize_v_quantile=0.995)
```
